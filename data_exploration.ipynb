```python
#Trying to find number of images in folder
#This proves that there are images which don't have defects

from pathlib import Path

folder = Path("data/raw/severstal-steel-defect-detection/train_images")

jpg_count = len(list(folder.glob("*.jpg")))

print(jpg_count) 
```

```python
#This includes trash files created while copying (One file has created one trash file)

from pathlib import Path

folder = Path("data/raw/severstal-steel-defect-detection/train_images")

file_count = sum(1 for f in folder.iterdir() if f.is_file())

print(file_count)
```

```python
#Deleting trash files from training images
from pathlib import Path

folder = Path("data/raw/severstal-steel-defect-detection/train_images")

for file in folder.iterdir():
    if file.is_file() and file.suffix.lower() != ".jpg":
        file.unlink()  # delete file
```

```python
#Sucessfully deleted trash files from training images.
from pathlib import Path
folder = Path("data/raw/severstal-steel-defect-detection/train_images")
file_count = sum(1 for f in folder.iterdir() if f.is_file())

print(file_count)
```

```python
#Deleting trash files from test images
from pathlib import Path

folder = Path("data/raw/severstal-steel-defect-detection/test_images")

for file in folder.iterdir():
    if file.is_file() and file.suffix.lower() != ".jpg":
        file.unlink()  # delete file
```
